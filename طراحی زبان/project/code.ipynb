## بررسی نوع انقیاد در زبان پایتون

### انقیاد دیر رس

```python
colors = []

for color in ["red", "green", "blue"]:
    def show():
        print(color)
    colors.append(show)

for b in colors:
    b()
```

### انقیاد زود رس

```python
colors = []

for color in ["red", "green", "blue"]:
    def show(color=color):
        print(color)
    colors.append(show)

for b in colors:
    b()
```

## تبدیل نوع در زبان پایتون

### تبدیل نوع ضمنی در پایتون

```python
A = 2
B = 5.3
print(A+B)
```

### تبدیل نوع صریح در پایتون

```python
A, B = 2, 5.3
print(float(A)+B)
```

### چالش های پشتیبانی زبان از هر دو تبدیل نوع

```python
A, B = True, 2
print(A+B)
```

## نوع شمارش در پایتون

```python
from enum import Enum

class colors(Enum):
    red = 1
    green = 2
    blue = 3
```

```python
for A, B in enumerate(["red", "green", "blue"]):
    print(A, B)
```

## انواع داده در پایتون

```python
A = 1
B = 0.1
C = "hello world"
D = [1, 2, 3]
E = (1, 2, 3)
F = {'name': 'reza'}
G = {1, 2, 3}
H = True
I = None

for i in [A, B, C, D, E, F, G, H, I]:
    print(type(i))
```

## کپسول سازی در پایتون

```python
class test():
    def __private(self):
        print("private")
    
    def public(self):
        print("public")
    
test1 = test()
test1.public()
test1.__private()
```
