## بررسی نوع انقیاد در زبان پایتون

### انقیاد دیر رس

```python
colors = []

for color in ["red", "green", "blue"]:
    def show():
        print(color)
    colors.append(show)

for b in colors:
    b()
```

### انقیاد زود رس

```python
colors = []

for color in ["red", "green", "blue"]:
    def show(color=color):
        print(color)
    colors.append(show)

for b in colors:
    b()
```
